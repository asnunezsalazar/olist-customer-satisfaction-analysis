# **3. Hypothesis Testing (A/B Testing)**

**Hipotesis: Testing the Effect of Shipping Time on Customer Satisfaction**

**Ho:** There is no difference in customer satisfaction between orders delivered on time and orders delivered late.

**H1:** Orders delivered on time have higher customer satisfaction compared to orders delivered late.

```python
df_test = orders.merge(orders_reviews, on="order_id")
```

```python
df_test["delivery_status"] = df_test.apply(
    lambda x: "On Time"
    if x["order_delivered_customer_date"] <= x["order_estimated_delivery_date"]
    else "Late",
    axis=1
)
```

```python
df_test["delivery_status"].value_counts()
```

```python
df_test.groupby("delivery_status")["review_score"].mean()
```

Customers whose orders are delivered on time report much higher satisfaction compared to the customers whose orders arrive late.

```python
from scipy.stats import ttest_ind

on_time = df_test[df_test["delivery_status"] == "On Time"]["review_score"]
late = df_test[df_test["delivery_status"] == "Late"]["review_score"]
```

```python
t_stat, p_value = ttest_ind(on_time, late, nan_policy="omit")

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

The p-value is smaller than 0.05, that means we reject the null hypothesis.

Also the value of the t-statistic = 156,6 tell us the statistical test becomes very reliable.

Conslusion: We can conclude that customers whose orders are delivered late are significantly less satisfied than those whose orders arrive on time.



**Using box plot for A/B testing**

We choose box plot to show the data because is the most common graph for A/B testing, also because is easier with this type of graph to see the central tendency of two groups and how the data is spread.

```python
fig = px.box(df_test, x="delivery_status", y="review_score", title="Customer Satisfaction: On-Time vs Late Deliveries")

fig.show()
```

**Checking the hypothesis in an histogram**

```python
fig = px.histogram(
    df_test,
    x="review_score",
    color="delivery_status",
    histnorm="probability density",
    barmode="overlay",
    nbins=20,
    title="Distribution of Customer Satisfaction Scores"
)

fig.show()
```

# **Conclusions of the A/B Testing**

This analysis examined the relationship between shipping performance and customer satisfaction in the Olist e-commerce dataset.

The results show a strong and statistically significant relationship between delivery time and customer satisfaction. Orders delivered on time received an average rating of 4.29, while late deliveries received an average rating of 2.35.

The hypothesis test confirmed that this difference is statistically significant (p-value < 0.05). Therefore, delivery delays have a substantial negative impact on customer satisfaction.
